```python
# Hospital Ward Staffing Optimizer

doctors = int(input("Enter number of doctors: "))
nurses = int(input("Enter number of nurses: "))
cleaners = int(input("Enter number of cleaners: "))
other_staff = int(input("Enter number of other staff: "))

total_staff = doctors + nurses + cleaners + other_staff

print("\n🏥 STAFFING SUMMARY")
print("-------------------------")
print("Doctors       :", doctors)
print("Nurses        :", nurses)
print("Cleaners      :", cleaners)
print("Other Staff   :", other_staff)
print("-------------------------")
print("Total Staff   :", total_staff)
```

```python
# Required staff according to ward size

beds = int(input("Enter number of beds in the ward: "))

required_doctors = max(1, beds // 10)
required_nurses = max(1, beds // 4)
required_cleaners = max(1, beds // 10)

print("\n📊 REQUIRED STAFF")
print("-------------------------")
print("Required Doctors  :", required_doctors)
print("Required Nurses   :", required_nurses)
print("Required Cleaners :", required_cleaners)

print("\n💡 STAFF STATUS")

if doctors >= required_doctors:
    print("Doctors: Sufficient")
else:
    print("Doctors: Shortage =", required_doctors - doctors)

if nurses >= required_nurses:
    print("Nurses: Sufficient")
else:
    print("Nurses: Shortage =", required_nurses - nurses)

if cleaners >= required_cleaners:
    print("Cleaners: Sufficient")
else:
    print("Cleaners: Shortage =", required_cleaners - cleaners)
```

```python
required_total = (
    required_doctors +
    required_nurses +
    required_cleaners
)

available_total = doctors + nurses + cleaners

print("\n🏥 WARD STAFFING OPTIMIZATION")
print("------------------------------")
print("Available Staff :", available_total)
print("Required Staff  :", required_total)

difference = available_total - required_total

if difference > 0:
    print("Status: Extra Staff =", difference)
elif difference < 0:
    print("Status: Staff Shortage =", abs(difference))
else:
    print("Status: Perfect Staffing")
```
